```python
import os

data_path = '/content/drive/MyDrive/CS114.P11.CarData/'
data_1 = ['Others']  # Thư mục chứa các thư mục con

for folder in data_1:
    folder_path = os.path.join(data_path, folder)
    if os.path.isdir(folder_path) and folder == 'Others':  # Kiểm tra nếu là thư mục "Others"
        # Duyệt qua tất cả các thư mục con trong thư mục "Others"
        car_brands = os.listdir(folder_path)
        for car_brand in car_brands:
            car_brand_path = os.path.join(folder_path, car_brand)
            if os.path.isdir(car_brand_path):  # Kiểm tra nếu là thư mục con (tên hãng xe)
                files = os.listdir(car_brand_path)
                # Đổi tên các tệp ảnh trong thư mục của mỗi hãng xe
                for i, file in enumerate(files):
                    # Chỉ đổi tên các tệp ảnh (bạn có thể thêm các định dạng khác nếu cần)
                    if file.endswith(('.jpg', '.png', '.jpeg')):
                        # Tạo tên mới cho tệp
                        new_name = f"22520459-22520507-22520862.{car_brand}.{i+1}.jpg"
                        old_file = os.path.join(car_brand_path, file)
                        new_file = os.path.join(car_brand_path, new_name)
                        os.rename(old_file, new_file)
                        print(f"Đã đổi tên {file} thành {new_name}")
```
