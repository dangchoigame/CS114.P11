```python
import os
import pandas as pd
from collections import defaultdict

# Đường dẫn tới thư mục chứa dữ liệu
data_path = '/content/drive/MyDrive/Public/'

# Khởi tạo dictionary để lưu số lượng ảnh của mỗi sinh viên và theo từng hiệu xe
student_image_count = defaultdict(float)
student_brand_count = defaultdict(lambda: defaultdict(float))

# Duyệt qua các thư mục con (mỗi thư mục con là một hãng xe)
data_1 = os.listdir(data_path)
for folder in data_1:
    folder_path = os.path.join(data_path, folder)

    if os.path.isdir(folder_path):  # Kiểm tra xem có phải thư mục không
        car_brand = folder  # Tên thư mục chính là thương hiệu xe
        files = os.listdir(folder_path)  # Lấy danh sách các tệp trong thư mục

        # Duyệt qua từng tệp trong thư mục và thống kê số lượng ảnh
        for file in files:
            if file.endswith(('.jpg', '.png', '.jpeg')):
                # Tên tệp theo định dạng đã đổi: MSSV-MSSV-MSSV.car_brand.index.jpg
                parts = file.split('.')  # Phân tách tên tệp để lấy thông tin MSSV và hiệu xe
                if len(parts) < 2:
                    continue

                file_name = parts[0]
                mssvs = file_name.split('-')  # Danh sách MSSV tham gia đóng góp

                if len(mssvs) < 2:
                    continue  # Nếu không có đủ sinh viên thì bỏ qua

                # Tính tỷ lệ đóng góp cho mỗi sinh viên (chia đều cho tất cả sinh viên)
                contribution = 1 / len(mssvs)  # Mỗi sinh viên đóng góp 1/n ảnh

                # Thống kê số lượng ảnh cho từng sinh viên
                for mssv in mssvs:
                    student_image_count[mssv] += contribution  # Cộng dồn ảnh cho sinh viên
                    student_brand_count[mssv][car_brand] += contribution  # Cộng dồn ảnh cho hiệu xe của sinh viên

# Tạo CarDataset-1.csv: Thống kê tổng số lượng ảnh của từng sinh viên
data1 = []
for mssv, count in student_image_count.items():
    data1.append([mssv, 'All', round(count, 1)])  # Làm tròn đến 1 chữ số thập phân

df1 = pd.DataFrame(data1, columns=["MSSV", "All", "Số lượng"])

# Tạo CarDataset-2.csv: Thống kê số lượng ảnh đóng góp theo từng hiệu xe
data2 = []
for mssv, brands in student_brand_count.items():
    for brand, count in brands.items():
        data2.append([mssv, brand, round(count, 1)])  # Làm tròn đến 1 chữ số thập phân

df2 = pd.DataFrame(data2, columns=["MSSV", "Hiệu xe", "Số lượng"])

# In ra nội dung của hai DataFrame trong notebook (dành cho kiểm tra kết quả)
print("\nCarDataset-1.csv - Thống kê tổng số ảnh đóng góp của từng sinh viên:")
print(df1)

print("\nCarDataset-2.csv - Thống kê số lượng ảnh đóng góp theo từng hiệu xe của sinh viên:")
print(df2)

# # Lưu các DataFrame vào file CSV
# df1.to_csv('/content/drive/MyDrive/CarDataset-1.csv', index=False, encoding='utf-8')
# df2.to_csv('/content/drive/MyDrive/CarDataset-2.csv', index=False, encoding='utf-8')

# # Thông báo lưu thành công
# print("\nĐã lưu CarDataset-1.csv và CarDataset-2.csv thành công vào MyDrive.")
```
