```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.preprocessing import image
import numpy as np
import joblib

# Load MobileNetV2 model và loại bỏ phần cuối (fully connected layers)
weights_path = '/kaggle/input/pretrained-weights/mobilenet_v2_weights_tf_dim_ordering_tf_kernels_1.0_224_no_top.h5'
base_model = MobileNetV2(weights=weights_path, include_top=False, input_shape=(224, 224, 3))

# Chuẩn bị ảnh
def preprocess_image(image_path):
    img = image.load_img(image_path, target_size=(224, 224))
    img_array = image.img_to_array(img)
    img_array /= 255.0  # Chuẩn hóa dữ liệu từ 0-1
    img_array = np.expand_dims(img_array, axis=0)  # Thêm chiều batch
    return img_array  # Trả về ảnh đã qua xử lý

# Dự đoán và trích xuất đặc trưng
image_path = '/kaggle/input/democar/vinfast.jpeg'
processed_image = preprocess_image(image_path)

# Trích xuất đặc trưng
try:
    features = base_model.predict(processed_image)
    # print(f"Hình dạng đặc trưng: {features.shape}")  # Kiểm tra hình dạng của đặc trưng
    features = features.flatten().reshape(1, -1)  # Làm phẳng và chuyển đổi thành ma trận 2D
    # print(f"Hình dạng đặc trưng: {features.shape}")  # Kiểm tra hình dạng của đặc trưng
except Exception as e:
    print(f"Lỗi dự đoán: {e}")

# Sử dụng các đặc trưng để dự đoán với mô hình trained khác
model_path = '/kaggle/input/rfmodel-car-nobatch/Final_RandomForest_Model.joblib'
rf_model = joblib.load(model_path)

# Dự đoán
try:
    predictions = rf_model.predict(features)
    print(predictions)
except Exception as e:
    print(f"Lỗi dự đoán mô hình RandomForest: {e}")
```
